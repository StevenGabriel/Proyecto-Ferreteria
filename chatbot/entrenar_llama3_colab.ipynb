## 🔑 Paso 2: Autenticación con Hugging Face y Carga del Modelo Base

```python
import torch
from unsloth import FastLanguageModel
from huggingface_hub import login

# Autenticación con tu token de Hugging Face
HF_TOKEN = "TU_TOKEN_DE_HUGGING_FACE_AQUI"
login(token=HF_TOKEN)

max_seq_length = 2048  # Longitud máxima de contexto
dtype = None           # Auto-detección (Float16 o Bfloat16)
load_in_4bit = True    # Cuantización de 4 bits para ahorrar VRAM

print("⏳ Cargando modelo base meta-llama/Llama-3.2-3B-Instruct...")
model, tokenizer = FastLanguageModel.from_pretrained(
    model_name = "unsloth/Llama-3.2-3B-Instruct-bnb-4bit",
    max_seq_length = max_seq_length,
    dtype = dtype,
    load_in_4bit = load_in_4bit,
)
print("✅ Modelo base cargado exitosamente en memoria GPU.")
```

## ⚙️ Paso 3: Configuración de Adaptadores LoRA (QLoRA)

```python
model = FastLanguageModel.get_peft_model(
    model,
    r = 16, # Rango LoRA
    target_modules = ["q_proj", "k_proj", "v_proj", "o_proj",
                      "gate_proj", "up_proj", "down_proj"],
    lora_alpha = 16,
    lora_dropout = 0, # Optimizado a 0 por Unsloth
    bias = "none",
    use_gradient_checkpointing = "unsloth",
    random_state = 3407,
    use_rslora = False,
    loftq_config = None,
)
print("✅ Adaptadores LoRA aplicados a la arquitectura Transformer.")
```

## 📚 Paso 4: Carga y Formateo del Dataset de Ferretería

```python
from datasets import load_dataset
from unsloth.chat_templates import get_chat_template

# Aplicar la plantilla oficial de chat de Llama 3.2
tokenizer = get_chat_template(
    tokenizer,
    chat_template = "llama-3.1",
)

def formatting_prompts_func(examples):
    convos = examples["messages"]
    texts = [tokenizer.apply_chat_template(convo, tokenize = False, add_generation_prompt = False) for convo in convos]
    return { "text" : texts }

# Cargar el dataset JSONL
dataset = load_dataset("json", data_files="dataset_ferreteria.jsonl", split="train")
dataset = dataset.map(formatting_prompts_func, batched = True)

print(f"✅ Dataset cargado con {len(dataset)} ejemplos formateados.")
print("\n--- Ejemplo del texto de entrenamiento ---")
print(dataset[0]["text"][:300] + "...")
```

## 🏋️‍♂️ Paso 5: Entrenamiento del Modelo (Fine-Tuning con SFTTrainer)

```python
from trl import SFTTrainer
from transformers import TrainingArguments
from unsloth import is_bfloat16_supported

trainer = SFTTrainer(
    model = model,
    tokenizer = tokenizer,
    train_dataset = dataset,
    dataset_text_field = "text",
    max_seq_length = max_seq_length,
    dataset_num_proc = 2,
    packing = False,
    args = TrainingArguments(
        per_device_train_batch_size = 2,
        gradient_accumulation_steps = 4,
        warmup_steps = 5,
        num_train_epochs = 3,
        learning_rate = 2e-4,
        fp16 = not is_bfloat16_supported(),
        bf16 = is_bfloat16_supported(),
        logging_steps = 1,
        optim = "adamw_8bit",
        weight_decay = 0.01,
        lr_scheduler_type = "linear",
        seed = 3407,
        output_dir = "outputs",
    ),
)

print("🚀 Iniciando entrenamiento...")
trainer_stats = trainer.train()
print("🎉 ¡Entrenamiento completado exitosamente!")
```

## 🧪 Paso 6: Prueba de Inferencia del Chatbot Afinado

```python
FastLanguageModel.for_inference(model) # Activa modo de inferencia 2x más rápido

# Pregunta de prueba sobre la ferretería
messages = [
    {"role": "system", "content": "Eres el asistente virtual inteligente de 'C&C Ferretería Casa y Construcción'."}, 
    {"role": "user", "content": "Hola, ¿qué descuentos dan en su programa de lealtad y qué cemento me recomiendan para una losa?"}
]

inputs = tokenizer.apply_chat_template(
    messages,
    tokenize = True,
    add_generation_prompt = True,
    return_tensors = "pt"
).to("cuda")

outputs = model.generate(input_ids = inputs, max_new_tokens = 256, use_cache = True, temperature = 0.4)
respuesta = tokenizer.batch_decode(outputs)
print(respuesta[0].split("<|start_header_id|>assistant<|end_header_id|>")[-1].replace("<|eot_id|>", ""))
```
